```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Generate random age data (centered around 35)
np.random.seed(42)
ages = np.random.normal(loc=35, scale=10, size=250).astype(int)

# Create a DataFrame
df = pd.DataFrame({'Age': ages})

# Plot the histogram
# 'bins' controls the number of bars, 'edgecolor' separates them visually
plt.hist(df['Age'], bins=15, edgecolor='black', color='skyblue')

# Add titles and labels
plt.title('Age Distribution of Participants')
plt.xlabel('Age Groups')
plt.ylabel('Number of People (Frequency)')

# Display the plot
plt.show()
```

```python
# TASK-1

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12, 15, 14, 20, 24, 22]

plt.plot(months, sales, marker='o')

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales (in thousands)")

plt.grid(True)
plt.show()

print("Month with highest sales: May")
```

```python
# TASK-2

import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

tips = px.data.tips()

avg_tip = tips.groupby("day")["tip"].mean().reindex(["Thur", "Fri", "Sat", "Sun"])

plt.bar(avg_tip.index, avg_tip.values, color="skyblue")

plt.title("Average Tip for Each Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")

plt.show()

print("Day with highest average tip:", avg_tip.idxmax())
```

```python
# TASK-3

import matplotlib.pyplot as plt
import plotly.express as px

tips = px.data.tips()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Histogram
axes[0].hist(tips["total_bill"], bins=10, color="skyblue")
axes[0].set_title("Histogram of Total Bill")
axes[0].set_xlabel("Total Bill")
axes[0].set_ylabel("Frequency")

# Scatter Plot
axes[1].scatter(tips["total_bill"], tips["tip"], color="green")
axes[1].set_title("Total Bill vs Tip")
axes[1].set_xlabel("Total Bill")
axes[1].set_ylabel("Tip")

plt.tight_layout()
plt.savefig("tips_plots.png", dpi=150)
plt.show()
```
