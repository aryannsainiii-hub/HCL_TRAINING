```python
#TASK-7

import pandas as pd
import numpy as np

df = pd.read_csv("students.csv")
Marks = df["Marks"]

print("Mean:", Marks.mean())
print("Median:", Marks.median())
print("Mode:", Marks.mode().tolist())
print("Variance:", Marks.var())
print("Standard Deviation:", Marks.std())
print("Minimum:", Marks.min())
print("Maximum:", Marks.max())


Q1 = Marks.quantile(0.25)
Q3 = Marks.quantile(0.75)
IQR = Q3 - Q1

Lower_Limit = Q1 - 1.5 * IQR
Upper_Limit = Q3 + 1.5 * IQR

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", Lower_Limit)
print("Upper Limit:", Upper_Limit)

Outliers = df[(Marks < Lower_Limit) | (Marks > Upper_Limit)]

print("\nOutliers:")
print(Outliers)
```

```python

# TASK-9

import numpy as np
from scipy import stats

Marks = df["Marks"].dropna()

Mean = Marks.mean()
Std = Marks.std()
N = len(Marks)

CI = stats.t.interval(
    confidence=0.95,
    df=N-1,
    loc=Mean,
    scale=Std / np.sqrt(N)
)

print("Mean Marks:", Mean)
print("95% Confidence Interval:", CI)

print("\nThis interval means we are 95% confident that the true average marks of the population lies within this range.")
```

```python
# TASK-10

from scipy import stats


H0 = "Average marks = 60"
H1 = "Average marks != 60"

print("Null Hypothesis (H0):", H0)
print("Alternative Hypothesis (H1):", H1)

Result = stats.ttest_1samp(df["Marks"].dropna(), 60)

print("\nT-statistic:", Result.statistic)
print("P-value:", Result.pvalue)

if Result.pvalue < 0.05:
    print("\nReject H0. There is sufficient evidence that the average marks differ from 60.")
else:
    print("\nFail to reject H0. There is insufficient evidence that the average marks differ from 60.")

```

```python
# TASK-11

from scipy import stats

CSE_Marks = df[df["Branch"] == "CSE"]["Marks"].dropna()
ECE_Marks = df[df["Branch"] == "ECE"]["Marks"].dropna()


print("Null Hypothesis (H0): Mean marks of CSE and ECE are equal.")
print("Alternative Hypothesis (H1): Mean marks of CSE and ECE are different.")

Result = stats.ttest_ind(CSE_Marks, ECE_Marks)

print("\nP-value:", Result.pvalue)

if Result.pvalue < 0.05:
    print("Reject H0. There is a significant difference in average marks between CSE and ECE students.")
else:
    print("Fail to reject H0. There is insufficient evidence of a significant difference in average marks between CSE and ECE students.")
```
