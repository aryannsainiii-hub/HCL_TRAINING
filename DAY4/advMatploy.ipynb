```python
import matplotlib.pyplot as plt

months = ["jan","feb","mar","apr"]
sales  = [100,150,120,200]
profit = [20,35,25,50]

fig, ax = plt.subplots(1,2,figsize=(10,4))

ax[0].plot(months,sales,marker="o")
ax[0].set_title("monthly sales")
ax[0].set_xlabel("months")
ax[0].set_ylabel("sales")

ax[1].bar(months,profit)
ax[1].set_title("monthly profit")
ax[1].set_xlabel("months")
ax[1].set_ylabel("profit")
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 200]
profit = [20, 35, 25, 50]

fig, ax1 = plt.subplots(figsize=(9, 5))

line1 = ax1.plot(
    months, sales,
    color="blue",
    marker='o',
    label="Sales"
)

ax1.set_xlabel("Month")
ax1.set_ylabel("Sales (units)", color="blue")

ax2 = ax1.twinx()

line2 = ax2.plot(
    months, profit,
    color="green",
    marker='o',
    label="profit"
)

ax2.set_xlabel("Month")
ax2.set_ylabel("profit", color="green")
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 200]
profit = [20, 35, 25, 50]

fig, ax1 = plt.subplots(figsize=(9, 5))

line1 = ax1.plot(
    sales, months,
    color="blue",
    marker='o',
    label="Sales"
)

ax1.set_xlabel("sales")
ax1.set_ylabel("months (units)", color="blue")

ax2 = ax1.twiny()

line2 = ax2.plot(
    profit, months,
    color="green",
    marker='o',
    label="profit"
)

ax2.set_xlabel("profit", color="green")
ax2.set_ylabel("months")
```

```python
import matplotlib.pyplot as plt

hours = [1,2,3,4,5]
celsius = [0,10,20,30,40]

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(hours,celsius,marker="o",color='blue')

ax.set_xlabel("time (hours)")
ax.set_ylabel("temp(c)",color='blue')

ax.tick_params(axis='y',labelcolor='blue')

def celsius_to_fahreneit(c):
    return (c*9/5)+32

def fehreneit_to_celsius(f):
    return (f-32)*5/9

secax = ax.secondary_yaxis(
    'right',
    functions=(celsius_to_fahreneit,fehreneit_to_celsius)
)

secax.set_ylabel("temperature(f)",color='red')

secax.tick_params(axis='y',labelcolor='red')

ax.set_title("temp in cel and fah")

fig.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

fig, ax = plt.subplots()

ax.plot(
    [1,2,3,4],
    [10,20,15,25],
    color = "blue"
)

ax.axhspan(12,18,color="orange",alpha=0.3)

line_handle = Line2D(
    [0],[0],
    color="blue",
    linewidth=2
)

shade_handle = Patch(
    facecolor="orange",
    alpha = 0.3
)

ax.legend(
    handles=[line_handle,shade_handle],
    labels=["observed value","target range"]
)

ax.set_title("chart with proxy legend enteties")
ax.set_xlabel("x values")
ax.set_ylabel("y values")
```

```python
import matplotlib.pyplot as plt
fig, ax1  = plt.subplots()

line1, = ax1.plot(
    [1,2,3,4],
    [10,20,15,25],
    color = "blue",
    label = "temperature"
)
ax1.set_xlabel("time")
ax1.set_ylabel("temperature(c)",color="blue")

ax2  = ax1.twinx()
line2, = ax2.plot(
    [1,2,3,4],
    [50,60,55,70],
    color = "red",
    linestyle = "--",
    label = "humidity"
)
ax2.set_ylabel("humidity(%)",color = "red")

handles1,labels1 = ax1.get_legend_handles_labels()
handles2,labels2 = ax2.get_legend_handles_labels()

ax1.legend(
    handles1 + handles2,
    labels1 + labels2,
    loc = "upper left"
)

ax1.set_title("temprature and humidity")
plt.show()

```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import StrMethodFormatter

fig, ax = plt.subplots()

ax.plot([1,2,3,4],[1000,2500,3500,5000])


ax.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))

plt.show()
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.plot([1,2,3,4],[10,20,35,25], marker="o")
ax.set_title("Click Anywhere om the Plot")

def on_click(event):
    if event.inaxes is not None:
        print("X Coordinate: ", event.xdata)
        print("Y Coordinate: ", event.ydata)
fig.canvas.mpl_connect("button_press_event", on_click)
plt.show()
```
