```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px

sales = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Laptops": [40, 42, 45, 38, 36, 35, 41, 48, 52, 85, 95, 60],
    "Phones": [55, 54, 58, 60, 57, 59, 62, 64, 66, 90, 98, 72],
    "Headphones": [20, 22, 21, 25, 24, 26, 25, 27, 30, 45, 52, 40]
})

sales
```

## Part A: The Starting Point

```python
# TASK-1: Default Chart

plt.figure(figsize=(10, 5))

plt.plot(sales["Month"], sales["Laptops"], label="Laptops")
plt.plot(sales["Month"], sales["Phones"], label="Phones")
plt.plot(sales["Month"], sales["Headphones"], label="Headphones")

plt.legend()
plt.savefig("before.png")
plt.show()
```

## Part A: The Starting Point

```python
# TASK-2: Consistent, Readable Colours

import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

colors = {
    "Laptops": "#0072B2",
    "Phones": "#D55E00",
    "Headphones": "#009E73"
}

plt.figure(figsize=(10, 5))

plt.plot(sales["Month"], sales["Laptops"],
         color=colors["Laptops"], linewidth=2.5, marker="o", label="Laptops")

plt.plot(sales["Month"], sales["Phones"],
         color=colors["Phones"], linewidth=2.5, marker="o", label="Phones")

plt.plot(sales["Month"], sales["Headphones"],
         color=colors["Headphones"], linewidth=2.5, marker="o", label="Headphones")

plt.title("Monthly Sales of Three Products")
plt.xlabel("Month")
plt.ylabel("Sales (Rs. thousands)")
plt.legend()
plt.tight_layout()
plt.show()
```

## Part C: Titles and Labels


```python
# TASK-3: Titles and Labels

import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

colors = {
    "Laptops": "#0072B2",
    "Phones": "#D55E00",
    "Headphones": "#009E73"
}

fig, ax = plt.subplots(figsize=(11, 5))

ax.plot(sales["Month"], sales["Laptops"],
        color=colors["Laptops"], linewidth=2.5, marker="o")

ax.plot(sales["Month"], sales["Phones"],
        color=colors["Phones"], linewidth=2.5, marker="o")

ax.plot(sales["Month"], sales["Headphones"],
        color=colors["Headphones"], linewidth=2.5, marker="o")

ax.text(11.2, sales["Laptops"].iloc[-1], "Laptops",
        color=colors["Laptops"], va="center")

ax.text(11.2, sales["Phones"].iloc[-1], "Phones",
        color=colors["Phones"], va="center")

ax.text(11.2, sales["Headphones"].iloc[-1], "Headphones",
        color=colors["Headphones"], va="center")

ax.set_title(
    "Phones Lead Sales, and Every Product Peaks in November",
    fontsize=14,
    fontweight="bold",
    loc="left"
)

ax.set_xlabel("Month")
ax.set_ylabel("Sales (Rs. thousands)")


ax.spines[["top", "right"]].set_visible(False)

ax.set_xlim(0, 11.8)
plt.tight_layout()
plt.show()
```

## Part D: Annotations and Highlights

```python
# TASK-4: Annotations and Highlights

import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

colors = {
    "Laptops": "#0072B2",
    "Phones": "#D55E00",
    "Headphones": "#009E73"
}

fig, ax = plt.subplots(figsize=(11, 6))

ax.plot(sales["Month"], sales["Laptops"],
        color=colors["Laptops"], linewidth=2.5, marker="o", label="Laptops")

ax.plot(sales["Month"], sales["Phones"],
        color=colors["Phones"], linewidth=2.5, marker="o", label="Phones")

ax.plot(sales["Month"], sales["Headphones"],
        color=colors["Headphones"], linewidth=2.5, marker="o", label="Headphones")

ax.axvspan(8.5, 10.5, color="orange", alpha=0.2)

ax.annotate(
    "Festive season peak",
    xy=(10, 98),
    xytext=(5.5, 90),
    arrowprops=dict(arrowstyle="->")
)

ax.axhline(60, linestyle="--", color="grey")

ax.text(0.2, 62, "Monthly target = 60",
        color="grey", fontsize=10)

ax.set_title(
    "Phones Lead Sales During the Festive Season",
    fontsize=14,
    fontweight="bold",
    loc="left"
)

ax.set_xlabel("Month")
ax.set_ylabel("Sales (Rs. thousands)")
ax.legend()

ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()

plt.savefig("after.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# TASK-5: Highlight One Bar

import matplotlib.pyplot as plt

max_index = sales["Phones"].idxmax()

bar_colors = ["lightgrey"] * len(sales)
bar_colors[max_index] = "crimson"

fig, ax = plt.subplots(figsize=(10, 5))

bars = ax.bar(
    sales["Month"],
    sales["Phones"],
    color=bar_colors
)

ax.bar_label(bars, padding=3)

ax.set_title(
    "Phone Sales Peak in November",
    fontsize=14,
    fontweight="bold",
    loc="left"
)

ax.set_xlabel("Month")
ax.set_ylabel("Sales (Rs. thousands)")

ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout()
plt.show()

print("Month with highest phone sales:", sales.loc[max_index, "Month"])
print("Highest phone sales:", sales.loc[max_index, "Phones"], "thousand rupees")
```

## Part E: Customizing a Plotly Chart


```python
# TASK-6: Interactive Plotly Chart

import plotly.express as px

colors = {
    "Laptops": "#0072B2",
    "Phones": "#D55E00",
    "Headphones": "#009E73"
}

long = sales.melt(
    id_vars="Month",
    var_name="Product",
    value_name="Sales"
)

fig = px.line(
    long,
    x="Month",
    y="Sales",
    color="Product",
    markers=True,
    color_discrete_map=colors,
    template="simple_white",
    title="Phones Lead All Year, and Every Product Peaks in November"
)

fig.update_traces(
    line_width=3,
    hovertemplate="%{x}: Rs. %{y}k"
)

fig.update_layout(
    yaxis_title="Sales (Rs. thousands)",
    xaxis_title="Month"
)

fig.add_annotation(
    x="Nov",
    y=98,
    text="Festive season peak",
    showarrow=True,
    arrowhead=2
)

fig.add_vrect(
    x0="Oct",
    x1="Nov",
    fillcolor="orange",
    opacity=0.2,
    line_width=0
)

fig.show()
```
