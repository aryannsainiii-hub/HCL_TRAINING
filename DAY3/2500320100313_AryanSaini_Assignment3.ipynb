```python
# TASK-4

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

tips = px.data.tips()

sns.countplot(data=tips, x="day")
plt.title("Number of Customers on Each Day")
plt.xlabel("Day")
plt.ylabel("Number of Customers")
plt.show()

sns.barplot(data=tips, x="day", y="total_bill")
plt.title("Average Bill on Each Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.show()

print("Day with most customers:", tips["day"].value_counts().idxmax())
print("Day with highest average bill:", tips.groupby("day")["total_bill"].mean().idxmax())
```

```python
# TASK-5

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

tips = px.data.tips()

sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Total Bill Distribution for Each Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()

print("Day with the highest median bill: Sunday (Sun)")
print("Dots beyond the whiskers represent outliers.")

sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Total Bill by Day and Gender")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()

print("Difference: The second plot compares the total bill distributions of male and female customers for each day.")
```

```python
# TASK-1: Ranked Bar Chart

import matplotlib.pyplot as plt
import plotly.express as px

gap = px.data.gapminder()
gap_2007 = gap[gap["year"] == 2007]

Top10 = gap_2007.nlargest(10, "pop")

Top10 = Top10.sort_values(by="pop", ascending=True)

Population_Millions = Top10["pop"] / 1_000_000

plt.barh(Top10["country"], Population_Millions)

plt.title("Top 10 Countries by Population in 2007")
plt.xlabel("Population (in millions)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

print("The two countries far ahead of the rest are China and India.")
```

```python
# TASK-2: Grouped Bar Chart

import matplotlib.pyplot as plt
import plotly.express as px

gap = px.data.gapminder()

g = gap[gap["year"].isin([1952, 2007])].groupby(
    ["continent", "year"]
)["lifeExp"].mean().unstack()

print("Average Life Expectancy by Continent:")
print(g)

g.plot(kind="bar", figsize=(10, 6))

plt.title("Average Life Expectancy by Continent (1952 vs 2007)")
plt.xlabel("Continent")
plt.ylabel("Average Life Expectancy (Years)")
plt.xticks(rotation=0)
plt.legend(title="Year")
plt.tight_layout()
plt.show()

Improvement = g[2007] - g[1952]

print("\nImprovement in Life Expectancy:")
print(Improvement)

print("\nContinent with the greatest improvement:", Improvement.idxmax())
```

## Part B: Line Chart

```python
# TASK-3: Change Over Time

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

gap = px.data.gapminder()

countries = gap[gap["country"].isin(["India", "China", "Brazil", "United States"])]

sns.lineplot(
    data=countries,
    x="year",
    y="lifeExp",
    hue="country",
    marker="o"
)

plt.title("Life Expectancy Over Time")
plt.xlabel("Year")
plt.ylabel("Life Expectancy (Years)")
plt.legend(title="Country")
plt.tight_layout()
plt.show()

print("Country with the lowest life expectancy in 1952: India")
print("Country that improved the fastest overall: China")
print("A line chart is suitable because it shows changes and trends over time, making it easy to compare countries.")
```

## Part C: Scatter Plots


```python
# TASK-4: Linear Scale vs Log Scale

import matplotlib.pyplot as plt
import plotly.express as px

gap = px.data.gapminder()
gap_2007 = gap[gap["year"] == 2007]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(gap_2007["gdpPercap"], gap_2007["lifeExp"])
axes[0].set_title("Linear Scale")
axes[0].set_xlabel("GDP per Capita (US Dollars)")
axes[0].set_ylabel("Life Expectancy (Years)")

axes[1].scatter(gap_2007["gdpPercap"], gap_2007["lifeExp"])
axes[1].set_xscale("log")
axes[1].set_title("Logarithmic Scale")
axes[1].set_xlabel("GDP per Capita (US Dollars)")
axes[1].set_ylabel("Life Expectancy (Years)")

plt.tight_layout()
plt.show()

print("The logarithmic scale is easier to read because it spreads out countries with lower GDP and makes the overall relationship clearer.")
```

## Task 5: Bubble chart

```python
# TASK-5: Bubble Chart

import plotly.express as px

gap = px.data.gapminder()
gap_2007 = gap[gap["year"] == 2007]

fig = px.scatter(
    gap_2007,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="continent",
    hover_name="country",
    log_x=True,
    size_max=50,
    title="GDP per Capita vs Life Expectancy (2007)"
)

fig.show()

print("Continent with most countries having low life expectancy: Africa")
print("Two countries with the largest bubbles: China and India")
```

## HEATMAPS

```python
# TASK-6: Correlation Heatmap

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

gap = px.data.gapminder()

corr = gap[["lifeExp", "pop", "gdpPercap"]].corr()

print("Correlation Matrix:")
print(corr)

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

print("\nMost strongly related pair: Life Expectancy and GDP per Capita.")
print("Pair with the weakest relationship: Population and GDP per Capita.")
```

```python
# TASK-7: Pattern Heatmap

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

gap = px.data.gapminder()

pivot = gap.pivot_table(
    index="continent",
    columns="year",
    values="lifeExp",
    aggfunc="mean"
)

print("Average Life Expectancy by Continent and Year:")
print(pivot)

plt.figure(figsize=(12, 5))
sns.heatmap(pivot, annot=True, fmt=".0f", cmap="YlGnBu")

plt.title("Average Life Expectancy Across Continents (1952–2007)")
plt.xlabel("Year")
plt.ylabel("Continent")
plt.tight_layout()
plt.show()

print("\nContinent with the lowest life expectancy throughout: Africa")
print("The colour changes most noticeably in later decades, especially from 1982 to 2007, as life expectancy increases across most continents.")
```

```python
# TASK-9: Fix a Poor Chart

import matplotlib.pyplot as plt
import plotly.express as px

gap = px.data.gapminder()
gap_2007 = gap[gap["year"] == 2007]

bad = gap_2007[gap_2007["continent"] == "Europe"].nlargest(8, "lifeExp")

print("Problems with the original chart:")
print("1. Too many different colours distract the viewer.")
print("2. The y-axis starts at 79 instead of 0, exaggerating differences.")
print("3. Country names may overlap and become difficult to read.")
print("4. The chart has no title.")
print("5. The x-axis and y-axis have no labels.")

bad = bad.sort_values(by="lifeExp", ascending=True)

plt.figure(figsize=(10, 6))
plt.barh(bad["country"], bad["lifeExp"], color="steelblue")

plt.xlim(0, 100)
plt.title("Top 8 European Countries by Life Expectancy (2007)")
plt.xlabel("Life Expectancy (Years)")
plt.ylabel("Country")
plt.tight_layout()
plt.show()

print("After the fix, the bars start at zero, so differences in life expectancy appear proportional and less exaggerated.")
```
