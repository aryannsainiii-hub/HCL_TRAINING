```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

```python
df = pd.DataFrame({
    "students": ["s01","s02","s03","s04","s05","s06",
    "s07","s08","s09","s10","s11","s12"],
    "branch" : ["cse","ece","cse","me","ece","cse","ce","me","ece","cse","ce","ece"],
    "study_hours":[3,2,5,2,4,6,3,4,5,2,6,3],
    "marks": [65,58,82,55,72,91,68,70,80,60,88,75]
})
display(df)
```

```python
plt.Figure(figsize=(7,4))

sns.countplot(
    data = df,
    x = "branch",
    order=["cse","ece","me","ce"],
    color="cornflowerblue"
)

plt.title("number of students in each branch")
plt.xlabel("branch")
plt.ylabel("number of students")
plt.tight_layout()
plt.show()
```

```python
df["study_group"]=df["study_hours"].apply(
    lambda hours: "3 hours or less" if hours<=3  else "more than 3 hours"
)
display(df[["students","branch","study_hours","study_group"]])
plt.Figure(figsize=(9,4))
sns.countplot(
    data = df,
    x = "branch",
    hue="study_group",
    order=["cse","ece","me","Ce"]
)
plt.title("student count and branch and study group")
plt.xlabel("branch")
plt.ylabel("number of students")
plt.legend(title="study_group")
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(7,4))

sns.barplot(
    data=df,
    x="branch",
    y="marks",
    order=["cse","ece","me","ce"],
    errorbar=None
)


plt.title("average marks by branch")
plt.xlabel("branch")
plt.ylabel("avegrage marks")
plt.ylim(0,100)
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(7,4))

sns.barplot(
    data=df,
    estimator=max,
    x="branch",
    y="marks",
    order=["cse","ece","me","ce"],
    errorbar=None
)


plt.title("max marks by branch")
plt.xlabel("branch")
plt.ylabel("max marks")
plt.ylim(0,100)
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="branch",
    y="marks",
    order=["cse","ece","me","ce"],
)


plt.title("distribution of student marks by branch")
plt.xlabel("branch")
plt.ylabel("marks(out of 100)")
plt.ylim(0,105)
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(8,5))

sns.boxplot(
    data=df,
    y="marks",
)

plt.title("distribution of student marks by branch")
plt.xlabel("branch")
plt.ylabel("marks(out of 100)")
plt.ylim(0,105)
plt.tight_layout()
plt.show()
```
