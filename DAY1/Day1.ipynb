```python
# Initial List of systolic blood pressure readings (in mmHg)
bp_readings = [120, 130, 140, 120, 110, 135, 150, 160, 125, 130]
# Remove a duplicate reading of 120 mmHg
bp_readings.remove(120)
print("After removing duplicate 120 mmHg:", bp_readings)
# Remove the third reading (e.g., faulty device reading) bp_readings.pop(2)
print("After removing third reading:", bp_readings)
# Delete the first reading manually (e.g., Logged before calibration) del bp_readings[0]
print("After deleting first reading:", bp_readings)
# Clear all readings for a new patient or new day bp_readings.clear()
print("After clearing all readings:", bp_readings)
```

```python
import re

with open("students.txt", "r") as file:
    lines = file.readlines()

d = {}
for line in lines:
    if "," in line:
        parts = line.strip().split(",")
        name = parts[0].strip()
        score_part = parts[1].strip()
        
        numbers = re.findall(r'\d+', score_part)
        if numbers:
            d[name] = int(numbers[0])

print(d)

marks = list(d.values())
if marks:
    avg_marks = sum(marks) / len(marks)
    print("avg marks :", avg_marks)

    highest = max(marks)
    print("highest scores :", highest)

    answer = list(filter(lambda name: d[name] > avg_marks, d))
    print(answer)
else:
    print("No valid student scores found in the file.")
```
