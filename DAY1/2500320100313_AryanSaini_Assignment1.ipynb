```python
import numpy as np
import pandas as pd
from scipy import stats

np.random.seed(42)
n = 40
df = pd.DataFrame({
    "Student_ID": range(1, n + 1),
    "Branch": np.random.choice(["CSE", "ECE", "ME", "CE"], n),
    "Study_Hours": np.round(np.random.uniform(1, 8, n), 1),
    "Attendance": np.random.randint(55, 100, n),
})
df["Marks"] = np.clip(
    np.round(30 + 5 * df["Study_Hours"] + 0.2 * df["Attendance"] + np.random.normal(0, 6, n)),
    0, 100
).astype(int)

df.head()
```

```python
# TASK 1

def get_grade(marks):
    if marks >= 75:
        return 'A'
    elif marks >= 60:
        return 'B'
    elif marks >= 40:
        return 'C'
    else:
        return 'D'

df["Grade"] = df["Marks"].apply(get_grade)

print(df.head())

```

```python
# TASK 2

Below75_Attendance = [marks for marks, attendance in zip(df["Marks"], df["Attendance"]) if attendance < 75]

Above70_Marks = list(filter(lambda x: x >70, df["Marks"]))

df["Marks"] = list(map(lambda x: min(x+5, 100), df["Marks"]))

print("marks of the students having attendance below 75:\n")
print(Below75_Attendance)
print("\nmarks above 70: ", Above70_Marks)
print("\n")
print(df.head())
```

```python
#TASK-3

def summary(*values, round_to=2):
    return {
        "count": len(values),
        "minimum": min(values),
        "maximum": max(values),
        "average": round(sum(values) / len(values), round_to)
    }

print("Summary of given values:")
print(summary(10, 20, 30))

print("\nSummary of students' marks:")
print(summary(*df["Marks"]))
```

```python
#TASK-4

df.to_csv("students.csv", index=False)

def load_data(filename):
    try:
        data = pd.read_csv(filename)
        print(data.head())
    except FileNotFoundError:
        print("File not found! Please check the file name.")

print("Loading correct file:")
load_data("students.csv")

print("\nLoading wrong file:")
load_data("wrong.csv")
```

```python
#task-5

Branch_Average = df.groupby("Branch")[["Marks", "Study_Hours"]].mean()

print("Average marks and study hours for each branch:")
print(Branch_Average)

print("\nBranch with highest average marks:")
print(Branch_Average["Marks"].idxmax())

Top5_Students = df.sort_values(by="Marks", ascending=False).head(5)

print("\nTop 5 students by marks:")
print(Top5_Students)
```

```python
#TASK-6

import numpy as np

Marks_Array = np.array(df["Marks"])

Mean = np.mean(Marks_Array)
Std = np.std(Marks_Array)

Z_Score = (Marks_Array - Mean) / Std

print("Z-scores of all students:")
print(Z_Score)

Count = np.sum(Z_Score > 1)

print("\nNumber of students with z-score greater than 1:", Count)
```
